# Dicionário de produtos

Fonte editável dos nicks de produtos e geração de `dicionario-produtos.csv`.

```python
from pathlib import Path
import csv

DICIONARIO_PRODUTOS = {
    '31021010': 'Ureia',
    '31022100': 'SAM',
    '31055900': 'NP',
    '31042090': 'MOP',
    '31054000': 'MAP',
    '31031100': 'Superfosfatos>35 %',
    '31031900': 'O.Superfosfatos',
    '31052000': 'NPK',
    '31023000': 'NAM',
    '31059090': 'NK',
    '31026000': 'Nit. de Cálcio',
    '31043010': 'SOP',
    '12019000': 'Soja',
    '38089329': 'Herbicidas',
    '38089199': 'Inseticidas',
    '38089299': 'O.fungicidas',
    '09011110': 'Café .n.t',
    '17011400': 'O.açúcares cana',
}
```

```python
destino = Path('dicionario-produtos.csv')
with destino.open('w', encoding='utf-8', newline='') as arquivo:
    escritor = csv.DictWriter(arquivo, fieldnames=('ncm', 'nick'))
    escritor.writeheader()
    escritor.writerows({'ncm': ncm, 'nick': nick} for ncm, nick in DICIONARIO_PRODUTOS.items())

print(f'{len(DICIONARIO_PRODUTOS)} produtos gravados em {destino}')
```
